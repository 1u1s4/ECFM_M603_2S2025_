````python
# -*- coding: utf-8 -*-
from dataclasses import dataclass
from typing import Optional, Iterable

@dataclass
class Nodo:
    """Nodo de un árbol binario."""
    valor: str
    izq: Optional['Nodo'] = None
    der: Optional['Nodo'] = None
````

```plaintext
       A
     /   \
    B     C
   / \     \
  D   E     F
```

````python
def arbol_ejemplo() -> Nodo:
    D = Nodo("D"); E = Nodo("E"); F = Nodo("F")
    B = Nodo("B", D, E)
    C = Nodo("C", None, F)
    A = Nodo("A", B, C)
    return A
````

````python
arbol_ejemplo()
````

````python
def myFunc():
  yield "Hello"
  yield 51
  yield "Good Bye"

x = myFunc()

for z in x:
  print(z)
````

````python
def preorden(u: Optional[Nodo]) -> Iterable[str]:
    """Visita nodo, luego subárbol izquierdo y derecho."""
    if u is None:
        return
    yield u.valor
    yield from preorden(u.izq)
    yield from preorden(u.der)

def inorden(u: Optional[Nodo]) -> Iterable[str]:
    """Visita izquierdo, luego nodo, luego derecho (sólo sentido en binarios)."""
    if u is None:
        return
    yield from inorden(u.izq)
    yield u.valor
    yield from inorden(u.der)

def postorden(u: Optional[Nodo]) -> Iterable[str]:
    """Visita subárbol izquierdo y derecho, y al final el nodo."""
    if u is None:
        return
    yield from postorden(u.izq)
    yield from postorden(u.der)
    yield u.valor
````

````python
def preorden_lista(u):
    if u is None:
        return []
    return [u.valor] + preorden_lista(u.izq) + preorden_lista(u.der)
````

````python
def preorden_traza(u: Optional[Nodo], prof: int = 0) -> None:
    if u is None:
        print("  " * prof + "⟲ retornar (nodo nulo)")
        return
    print("  " * prof + f"→ visitar {u.valor}")
    print("  " * prof + "↙ izq")
    preorden_traza(u.izq, prof + 1)
    print("  " * prof + "↘ der")
    preorden_traza(u.der, prof + 1)

def inorden_traza(u: Optional[Nodo], prof: int = 0) -> None:
    if u is None:
        print("  " * prof + "⟲ retornar (nodo nulo)")
        return
    print("  " * prof + "↙ izq")
    inorden_traza(u.izq, prof + 1)
    print("  " * prof + f"→ visitar {u.valor}")
    print("  " * prof + "↘ der")
    inorden_traza(u.der, prof + 1)

def postorden_traza(u: Optional[Nodo], prof: int = 0) -> None:
    if u is None:
        print("  " * prof + "⟲ retornar (nodo nulo)")
        return
    print("  " * prof + "↙ izq")
    postorden_traza(u.izq, prof + 1)
    print("  " * prof + "↘ der")
    postorden_traza(u.der, prof + 1)
    print("  " * prof + f"→ visitar {u.valor}")
````

````python
raiz = arbol_ejemplo()

print("Árbol de ejemplo: A(B(D,E), C(_,F))\n")
````

````python
print("Preorden :", list(preorden_lista(raiz)))   # Esperado: A B D E C F
````

````python
print("Preorden :", list(preorden(raiz)))   # Esperado: A B D E C F
print("Inorden  :", list(inorden(raiz)))    # Esperado: D B E A C F
print("Postorden:", list(postorden(raiz)))  # Esperado: D E B F C A
````

````python
print("\n=== Traza PREORDEN ===")
preorden_traza(raiz)
````

````python
print("\n=== Traza INORDEN ===")
inorden_traza(raiz)
````

````python
print("\n=== Traza POSTORDEN ===")
postorden_traza(raiz)
````
